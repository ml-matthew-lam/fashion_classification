```python
import matplotlib.pyplot as plt
from dataset import train_loader
```

```python
# test the train_loader: get it to deliver one batch of 64 images + labels
images, labels = next(iter(train_loader))

print("images shape:", images.shape)
print("labels shape:", labels.shape)
```

```python
# dictionnary that maps class numbers (used in dataset) to english class names
labels_map = {0: "t-shirt", 1: "trouser", 2: "pullover", 3: "dress", 4: "coat", 5: "sandal", 6: "shirt", 7: "sneaker", 8: "bag", 9: "ankle boot"}
```

```python
# visual check: show a grid of first 16 clothing items with their corresponding class labels
fig = plt.figure(figsize = (8,8))
for i in range(16):
    fig.add_subplot(4, 4, i+1)
    plt.imshow(images[i].squeeze(), cmap = "gray")
    plt.title(labels_map[labels.tolist()[i]])
    plt.axis("off")
plt.tight_layout()
plt.show()
```
